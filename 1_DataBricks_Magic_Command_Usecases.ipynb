```python
%sql
CREATE VOLUME IF NOT EXISTS workspace.default.usage_metrics;
```

```python
%python
import requests

URl="https://public.tableau.com/app/sample-data/mobile_os_usage.csv"
r=requests.get(URl)
display(r.text)
dbutils.fs.put("/Volumes/workspace/default/usage_metrics/mobile_os_usage.csv",r.text,overwrite=True)
```

```python
df=spark.read.csv("/Volumes/workspace/default/usage_metrics/mobile_os_usage.csv",inferSchema=True,header=True)
display(df)
```

```python
%fs
ls  /Volumes/workspace/default/usage_metrics
```

```python
%python
df=(df.withColumnRenamed("Mobile Operating System","Mobile_Operating_System").withColumnRenamed("Percent of Usage","Percent_of_Usage"))
display(df)
```

```python
%python
df.write.format("delta").mode("overwrite").saveAsTable("default.mobile_os_usage")
```
