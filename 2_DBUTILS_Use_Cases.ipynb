####3.Widget is used to parameterization and dynamic way to pass the value 


```python
print(" create a textbox widget")
dbutils.widgets.text("Country","USA","Enter the country name")
```

```python
print("test the widget")
country_name=dbutils.widgets.get("Country")
print("The country that is given as input is ",country_name)
```

```python
dbutils.widgets.remove("Country")
```

```python
dbutils.widgets.multiselect("multiselect_widget","wd36",["wd32","we43","we45","wd36",'wd38','we49'],"Select your team name")
team_name=dbutils.widgets.get("multiselect_widget")
print(type(team_name))
con_list=team_name.split(",")
print(type(con_list))
print(con_list[0])
list_to_str=",".join(con_list)
print(list_to_str)
print(con_list[1])
```

-Combo box is used to do multi select and also type the value you used 

```python
dbutils.widgets.combobox("Country_Combobox","USA",["CANADA","INDIA","MEXICO"],"Enter the coutry name")
country_name=dbutils.widgets.get("Country_Combobox")
print(country_name)
```
